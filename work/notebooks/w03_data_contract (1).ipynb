```python
# Load April 2026 for the future outcome window

april_path = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="fact_content_daily_performance/month=2026-04/data_0.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

april = pd.read_parquet(april_path)

print("April shape:", april.shape)
print("April dates:", april["report_date"].min(), "to", april["report_date"].max())
```

```python
# Build a simple future-performance proxy from April

future = (
    april.groupby(["client_hash_id", "content_hash_id"], as_index=False)
    .agg(
        future_impressions=("gsc_impressions", "sum"),
        future_clicks=("gsc_clicks", "sum")
    )
)

print("Future outcome rows:", len(future))
display(future.head())
```

```python
# Create the future-performance label
# 1 = declining, 0 = not declining

future["is_declining_label"] = (
    future["future_impressions"] == 0
).astype(int)

print("Label distribution:")
print(future["is_declining_label"].value_counts())
```

### Data limits

### Five features — available when?

* `gsc_impressions` — knowable at the decision moment because it comes from the observed GSC performance available for that report date.
* `gsc_clicks` — knowable at the decision moment because it comes from the observed GSC performance available for that report date.
* `gsc_avg_position` — knowable at the decision moment because it is an observed GSC ranking signal for that report date.
* `ga4_sessions` — knowable at the decision moment when GA4 data is available for that row.
* `scroll_events` — knowable at the decision moment when the corresponding GA4 event data is available for that row.


```python
# Build the future outcome from April 2026

april_path = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="fact_content_daily_performance/month=2026-04/data_0.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

april = pd.read_parquet(april_path)

print("April shape:", april.shape)
print("April dates:", april["report_date"].min(), "to", april["report_date"].max())

# Future performance proxy
future = (
    april.groupby(
        ["client_hash_id", "content_hash_id"],
        as_index=False
    )
    .agg(
        future_impressions=("gsc_impressions", "sum"),
        future_clicks=("gsc_clicks", "sum")
    )
)

# 1 = declining, 0 = not declining
future["is_declining_label"] = (
    future["future_impressions"] == 0
).astype(int)

print("Future outcome rows:", len(future))
print("Label distribution:")
print(future["is_declining_label"].value_counts())
```

### Deliberate leakage experiment

```python
from google.colab import userdata
from huggingface_hub import hf_hub_download
import pandas as pd

HF_TOKEN = userdata.get("HF_TOKEN")

file_path = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="fact_content_daily_performance/month=2026-03/data_0.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

march = pd.read_parquet(file_path)

print("March shape:", march.shape)
```

```python
# Deliberate leakage experiment
# Intentionally bring the future label into the March feature data.

leaky_features = march[
    [
        "client_hash_id",
        "content_hash_id",
        "gsc_impressions",
        "gsc_clicks",
        "gsc_avg_position",
        "ga4_sessions",
        "scroll_events",
    ]
].merge(
    future[
        [
            "client_hash_id",
            "content_hash_id",
            "is_declining_label",
        ]
    ],
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

print("Leaky feature frame shape:", leaky_features.shape)
print("Leakage column added:", "is_declining_label" in leaky_features.columns)

display(leaky_features.head())
```

### Leakage lesson

The `is_declining_label` column was deliberately added to the feature frame even though it is derived from future April performance. This is leakage because the label would not be available at the March decision moment. The experiment shows why future outcome information must not be used as an input feature.

The leakage column is removed from the final feature set.

```python
# Remove the deliberately leaked future label.
honest_features = leaky_features.drop(
    columns=["is_declining_label"]
)

print("Final honest feature columns:")
print(honest_features.columns.tolist())

print("Final feature count:", len(honest_features.columns))
```
